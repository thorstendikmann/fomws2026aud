Let's check if all packages from `requirements.txt` are installed:

```python
import importlib.util
import pathlib

requirements_path = pathlib.Path("../../../requirements.txt").resolve()

missing = []
with open(requirements_path) as f:
    for line in f:
        line = line.strip()
        if not line or line.startswith("#"):
            continue
        package = line.split("==")[0].strip()
        if importlib.util.find_spec(package) is None:
            missing.append(package)

if missing:
    raise ImportError(f"Missing packages: {', '.join(missing)}")
else:
    print("All packages from requirements.txt are installed.")
```
