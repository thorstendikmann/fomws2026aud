# Interactive 2D Collision Detection with a Quadtree

This notebook demonstrates how a quadtree reduces the number of object pairs that need an exact collision test.

Each moving object is a circle. At every simulation step:

1. Objects move and bounce at the borders.
2. A quadtree is rebuilt from their current positions.
3. Each object queries only nearby quadtree regions.
4. Candidate pairs receive an exact circle-to-circle collision test.

Circles show one object ID. Every quadtree region has an `N#` node label; `depth` is shown separately and means the level of that node in the tree. The text panel lists which object IDs each region contains. The hierarchy diagram shows parent-child links by depth. In the collision chart, teal dashed lines are candidate checks and red lines are actual intersections. The event log traces each circle through the inspected tree nodes, stored objects, candidate objects, and exact checks.
